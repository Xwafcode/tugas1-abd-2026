# Data Cleaning

Melakukan pembersihan data yang hilang dan anomali menggunakan Polars, kemudian menyimpannya ke format Parquet.

```python
import os
import polars as pl

DATA_PATH = '../data/modis_flood_features.csv'
CLEAN_DATA_PATH = '../data/modis_flood_features_clean.parquet'

if not os.path.exists(DATA_PATH):
    DATA_PATH = 'TUGAS1/data/modis_flood_features.csv'
    CLEAN_DATA_PATH = 'TUGAS1/data/modis_flood_features_clean.parquet'
```

```python
# Load data lazily
df = pl.scan_csv(DATA_PATH, ignore_errors=True)

# 1. Menghapus baris yang nilai lat atau lon-nya null karena krusial untuk analisis geospasial
# 2. Mengisi missing values pada precip_3d dengan median atau 0
df_clean = df.drop_nulls(subset=['lat', 'lon']).with_columns([
    pl.col('precip_3d').fill_null(0),
    pl.col('elevation').fill_null(0)
])

print(df_clean.collect_schema())
```

```python
# Mengeksekusi pipeline dan menyimpan hasilnya dalam format Parquet untuk efisiensi
df_clean.sink_parquet(CLEAN_DATA_PATH)
print("Data cleaning selesai dan disimpan ke format Parquet.")
```
