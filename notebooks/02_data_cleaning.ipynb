# Data Cleaning

Melakukan pembersihan data yang hilang dan anomali menggunakan Polars, kemudian menyimpannya ke format Parquet.

```python
import polars as pl

DATA_PATH = '../data/indonesia_sales_data.csv'
CLEAN_DATA_PATH = '../data/indonesia_sales_data_clean.parquet'
```

```python
# Load data lazily
df = pl.scan_csv(DATA_PATH)

# 1. Mengisi missing values pada province dengan 'Tidak Diketahui'
# 2. Mengisi missing values pada product_category dengan 'Lainnya'
# 3. Menyaring data dengan nilai penjualan negatif (jika ada)
df_clean = df.with_columns([
    pl.col('province').fill_null('Tidak Diketahui'),
    pl.col('product_category').fill_null('Lainnya')
]).filter(pl.col('total_sales') >= 0)

print(df_clean.collect_schema())
```

```python
# Mengeksekusi pipeline dan menyimpan hasilnya dalam format Parquet untuk efisiensi
df_clean.sink_parquet(CLEAN_DATA_PATH)
print("Data cleaning selesai dan disimpan ke format Parquet.")
```
