# Data Cleaning

Melakukan pembersihan data yang hilang dan anomali menggunakan Polars, kemudian menyimpannya ke format Parquet.

```python
import polars as pl

DATA_PATH = '../data/indonesia_housing_data.csv'
CLEAN_DATA_PATH = '../data/indonesia_housing_data_clean.parquet'
```

```python
# Load data lazily
df = pl.scan_csv(DATA_PATH, ignore_errors=True)

# 1. Mengisi missing values pada province
# 2. Mengisi missing values pada developer_name
# 3. Menyaring data dengan harga negatif atau tidak masuk akal
df_clean = df.with_columns([
    pl.col('province').fill_null('Tidak Diketahui'),
    pl.col('developer_name').fill_null('Tidak Diketahui')
]).filter(pl.col('price_idr') > 0)

print(df_clean.collect_schema())
```

```python
# Mengeksekusi pipeline dan menyimpan hasilnya dalam format Parquet untuk efisiensi
df_clean.sink_parquet(CLEAN_DATA_PATH)
print("Data cleaning selesai dan disimpan ke format Parquet.")
```
